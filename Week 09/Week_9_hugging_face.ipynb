# Sentiment Analysis

```python
from transformers import pipeline

sentiment_classifier = pipeline('sentiment-analysis')
```

```python
texts = ['This was a poignant film.',
         'It was a wonderful acting by the hero, that made the file look Good.',]

for text in texts:
  result = sentiment_classifier(text)
  print(f"Text: {text}")
  print(f"Result: {result}\n")
```

## Sentiment Analysis with specific model

```python
model_name = 'cardiffnlp/twitter-roberta-base-sentiment'

sentiment_classifier_specific = pipeline('sentiment-analysis',model=model_name)
```

```python
texts = ['This was a poignant film.',
         'It was a wonderful acting by the hero, that made the file look Good.',]

for text in texts:
  result = sentiment_classifier_specific(text)
  print(f"Text: {text}")
  print(f"Result: {result}\n")
```

```python
print(sentiment_classifier_specific.model.config.id2label)
```

# NER [Named Entity Recognitin]

```python
# The default model is 'dbmdz/bert-large-cased-finetuned-conll03-english'
ner_pipeline = pipeline('ner',aggregation_strategy='simple')
```

```python
texts = ["My name is S Ramanujam, and I am from Chennai.",
         "Barack Obama was born in Hawaii and served as President of the United States."]

for text in texts:
  entities = ner_pipeline(text)
  for entity in entities:
    print(f"  - Entity: {entity['word']}, Label: {entity['entity_group']}, Score: {entity['score']:.4f}")
```

## Specific model for NER task

```python
model_name = "dbmdz/bert-large-cased-finetuned-conll03-english"
ner_pipeline_specific = pipeline("ner",model=model_name,aggregation_strategy='simple')
```

```python
texts = ["My name is S Ramanujam, and I am from Chennai.",
         "Barack Obama was born in Hawaii and served as President of the United States."]

for text in texts:
  entities = ner_pipeline(text)
  for entity in entities:
    print(f"  - Entity: {entity['word']}, Label: {entity['entity_group']}, Score: {entity['score']:.4f}")
```

# Machine Translation

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "Helsinki-NLP/opus-mt-en-de"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = "The quick brown fox jumps over the lazy dog."

# Convert text → tokens
inputs = tokenizer(text, return_tensors="pt")

# Generate German translation
outputs = model.generate(**inputs)

# Convert tokens → text
translation = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("\n--- Specific Model (Helsinki-NLP/opus-mt-en-de) ---")
print(f"Original (EN): {text}")
print(f"Translation (DE): {translation}")
```
